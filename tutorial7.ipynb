```python
from scipy.special import comb

def p(x):

    c1 = comb(23, x)
    c2 = comb(18, 36 - x)
    c3 = comb(41, 36)

    return c1*c2/c3

p_list = []
for x in range(18, 24):

    p_list.append(p(x))

p_list = np.array(p_list)
print(p_list)
```
