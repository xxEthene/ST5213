```python
# General social survey
nij = np.array([[178, 138, 108], [570, 648, 442], [138, 252, 252]])
print(nij)
muhat = np.outer(nij.sum(axis=1), nij.sum(axis=0)) / nij.sum()
print(muhat)

X2 = ((nij - muhat)**2 / muhat).sum()
print(X2, chi2.sf(X2, 4))

G2 = 2 * (nij * np.log(nij/muhat)).sum()
print(G2, chi2.sf(G2, 4))
```

```python
# Hypergeometric distribution 
M = 8 # Total number of cups 
n = 4 # Total number of cups with milk poured first
N = 4 # Sample size drawn

print(hypergeom.pmf(range(5), M=M, n=n, N=N))
print(hypergeom.sf(range(-1, 4), M=M, n=n, N=N))
```

```python
# Fisher's exact test 
nij = np.array([3,1,1,3])                      # n_11 = 3
X2 = ((nij - 2)**2/2).sum()
print(chi2.sf(X2, 1))
```

```python
# Jane Austen 
df = pd.read_csv('dataset/JaneAusten.txt', sep=' ')
print(df.head(), '\n')

# convert Book and Word into categorical variables
df['Book'] = pd.Categorical(df['Book'])
df['Word'] = pd.Categorical(df['Word'])

print(df.info(), '\n')
tab = df.pivot(index='Word', columns='Book', values='Count').fillna(0)
print(tab)
```
